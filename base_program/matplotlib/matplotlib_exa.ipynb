```python
# 启用 matplotlib 在 notebook 中绘图
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# ==========================================
# 全局设置 (让图表更美观，支持中文)
# ==========================================
# 注意：如果本地没有 'SimHei'，可能会回退到默认字体。
# 这里尝试设置常用中文字体，防止乱码
plt.style.use('seaborn-v0_8') # 使用较新的 seaborn 风格
plt.rcParams['font.sans-serif'] = ['SimHei', 'Arial Unicode MS', 'Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False # 解决负号显示问题

# ==========================================
# 准备模拟数据 (无需外部文件)
# ==========================================
x = np.linspace(0, 10, 100)          # 连续数据
y_sin = np.sin(x)                    # 正弦波
y_cos = np.cos(x)                    # 余弦波

categories = ['A组', 'B组', 'C组', 'D组', 'E组']
values = [23, 45, 56, 78, 32]
values_2 = [12, 30, 45, 60, 20]

print("✅ 环境与数据准备完毕！请运行下方的具体图表单元格。")
```

# 📉 选项 A: 折线图 (Line Plot)

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(x, y_sin, label='正弦', color='blue', linewidth=2)
ax.plot(x, y_cos, label='余弦', color='red', linestyle='--')

ax.set_title('趋势分析：正弦与余弦')
ax.set_xlabel('时间/步长')
ax.set_ylabel('振幅')
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

# 📊 选项 B: 柱状图 (Bar Plot)

```python
fig, ax = plt.subplots(figsize=(10, 6))

x_pos = np.arange(len(categories))

# 绘制两组并列柱子
ax.bar(x_pos, values, color='skyblue', label='2023年', width=0.4)
ax.bar(x_pos + 0.4, values_2, color='salmon', label='2024年', width=0.4)

ax.set_xticks(x_pos + 0.2)
ax.set_xticklabels(categories)
ax.set_title('类别对比：两年数据')
ax.legend()
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

# 🥧 选项 C: 饼图 (Pie Chart)

```python
fig, ax = plt.subplots(figsize=(8, 8))

# 修正：pie 方法直接在 ax 上调用，explode 用法修正
wedges, texts, autotexts = ax.pie(
    values,
    labels=categories,
    autopct='%1.1f%%',
    startangle=90,
    explode=[0, 0.1, 0, 0, 0], # 把第二个块炸开
    shadow=True
)

# 美化字体
for text in texts:
    text.set_fontsize(12)
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontweight('bold')

ax.set_title('市场份额占比')

plt.tight_layout()
plt.show()
```

# 🌫️ 选项 D: 散点图 (Scatter Plot)

```python
fig, ax = plt.subplots(figsize=(10, 6))

np.random.seed(42)
rand_x = np.random.rand(50) * 10
rand_y = np.random.rand(50) * 10 + rand_x * 0.5

# c=rand_y 表示颜色随 Y 值变化，cmap 指定配色方案
scatter = ax.scatter(rand_x, rand_y, c=rand_y, cmap='viridis', alpha=0.6, edgecolors='w', s=100)

ax.set_title('相关性分析：X与Y的关系')
ax.set_xlabel('特征 X')
ax.set_ylabel('特征 Y')

# 添加颜色条
plt.colorbar(scatter, ax=ax, label='Y 值强度')
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

# 📦 选项 E: 箱线图 (Box Plot)

```python
fig, ax = plt.subplots(figsize=(10, 6))

data_to_plot = [np.random.normal(0, std, 100) for std in range(1, 4)]

# 绘制箱线图
bp = ax.boxplot(data_to_plot, labels=['低方差', '中方差', '高方差'], patch_artist=True)

# 给箱子填色
colors = ['lightblue', 'lightgreen', 'lightcoral']
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)
    patch.set_edgecolor('black')

ax.set_title('数据分布与异常值检测')
ax.grid(axis='y', alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()
```

# 🌊 选项 F: 直方图 (Histogram)

```python
fig, ax = plt.subplots(figsize=(10, 6))

data_hist = np.random.randn(1000)

ax.hist(data_hist, bins=30, color='purple', alpha=0.7, edgecolor='black')

ax.set_title('数据频率分布 (直方图)')
ax.set_xlabel('数值区间')
ax.set_ylabel('频数')
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

# 🗺️ 选项 G: 热力图 (Heatmap)

```python
fig, ax = plt.subplots(figsize=(8, 6))

# 生成随机相关矩阵 (5x5)
np.random.seed(0)
data_matrix = np.random.rand(5, 5)
# 为了更像相关系数，让它对称
corr_matrix = np.corrcoef(np.random.randn(5, 100))

im = ax.imshow(corr_matrix, cmap='coolwarm', aspect='auto', vmin=-1, vmax=1)

ax.set_xticks(range(5))
ax.set_yticks(range(5))
ax.set_xticklabels([f'Var{i}' for i in range(5)])
ax.set_yticklabels([f'Var{i}' for i in range(5)])
ax.set_title('相关性热力图')

# 添加颜色条
plt.colorbar(im, ax=ax, shrink=0.8)

plt.tight_layout()
plt.show()
```

# 📈 选项 H: 面积图 (Area Plot)

```python
fig, ax = plt.subplots(figsize=(10, 6))

# fill_between 用于填充两条线之间的区域
ax.fill_between(x, y_sin, y_cos, where=(y_sin > y_cos), interpolate=True, color='green', alpha=0.3, label='正弦 > 余弦')
ax.fill_between(x, y_sin, y_cos, where=(y_sin <= y_cos), interpolate=True, color='red', alpha=0.3, label='正弦 <= 余弦')

ax.plot(x, y_sin, 'k-', lw=1, label='正弦线')
ax.plot(x, y_cos, 'k-', lw=1, label='余弦线')

ax.set_title('区域填充：差异可视化')
ax.set_xlabel('时间')
ax.set_ylabel('振幅')
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

# 🚀 进阶挑战：多子图对比

```python
# 创建 1行2列 的画布
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# --- 左图：折线图 ---
ax1.plot(x, y_sin, color='blue', linewidth=2, label='正弦')
ax1.plot(x, y_cos, color='red', linestyle='--', label='余弦')
ax1.set_title('左图：趋势变化')
ax1.set_xlabel('时间')
ax1.set_ylabel('振幅')
ax1.legend()
ax1.grid(True, alpha=0.3)

# --- 右图：直方图 ---
data_right = np.random.randn(1000)
ax2.hist(data_right, bins=30, color='green', alpha=0.7, edgecolor='black')
ax2.set_title('右图：数据分布')
ax2.set_xlabel('数值')
ax2.set_ylabel('频数')
ax2.grid(axis='y', alpha=0.3)

plt.suptitle('综合练习：多子图展示', fontsize=16, fontweight='bold')
plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # 为总标题留出空间
plt.show()
```
